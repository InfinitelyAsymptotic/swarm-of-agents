# Introduction to Multi-Agent Systems

## What are Multi-Agent Systems?

Multi-agent systems (MAS) are computational systems where multiple autonomous agents interact to solve complex problems that are beyond the capability of a single agent.

### Key Concepts:

- **Agent**: An autonomous entity that can perceive, reason, and act
- **Autonomy**: Agents operate independently without direct human intervention
- **Communication**: Agents exchange messages and information
- **Coordination**: Agents work together to achieve common goals
- **Orchestration**: The design and management of agent interactions

## Why Multi-Agent Systems?

### Advantages:

1. **Specialization**: Each agent can focus on specific tasks
2. **Scalability**: Add more agents to handle increased workload
3. **Robustness**: System continues if one agent fails
4. **Parallel Processing**: Multiple agents work simultaneously
5. **Modularity**: Easy to update or replace individual agents

### Use Cases:

- Software development (planning, coding, testing, reviewing)
- Research and analysis (data collection, processing, synthesis)
- Customer service (routing, answering, escalation)
- Content creation (research, writing, editing, fact-checking)

## Example 2: Building a Simple Agent

Let's implement a basic agent class to understand the core components:

```python
from typing import Dict, List, Any
from dataclasses import dataclass
from datetime import datetime

@dataclass
class Message:
    """Represents a message between agents."""
    sender: str
    recipient: str
    content: str
    timestamp: datetime = None
    
    def __post_init__(self):
        if self.timestamp is None:
            self.timestamp = datetime.now()


class Agent:
    """Base class for an autonomous agent."""
    
    def __init__(self, name: str, role: str, capabilities: List[str]):
        self.name = name
        self.role = role
        self.capabilities = capabilities
        self.message_history: List[Message] = []
        
    def perceive(self, message: Message) -> None:
        """Receive and store incoming messages."""
        self.message_history.append(message)
        print(f"[{self.name}] Received: {message.content[:50]}...")
        
    def reason(self, task: str) -> str:
        """Process information and decide on actions."""
        # Simplified reasoning
        return f"[{self.name}] Processing task: {task}"
        
    def act(self, task: str) -> Message:
        """Perform an action and generate output."""
        result = self.reason(task)
        return Message(
            sender=self.name,
            recipient="system",
            content=result
        )
    
    def __repr__(self):
        return f"Agent(name='{self.name}', role='{self.role}')"


# Create sample agents
researcher = Agent(
    name="Researcher",
    role="Information Gathering",
    capabilities=["web_search", "document_analysis", "data_extraction"]
)

planner = Agent(
    name="Planner",
    role="Task Planning",
    capabilities=["task_decomposition", "scheduling", "resource_allocation"]
)

print(f"Created agents:")
print(f"  - {researcher}")
print(f"  - {planner}")
print(f"\nResearcher capabilities: {researcher.capabilities}")
print(f"Planner capabilities: {planner.capabilities}")
```

## Example 3: Agent Interaction

Let's see how agents communicate:

```python
# Create a message from researcher to planner
research_results = Message(
    sender="Researcher",
    recipient="Planner",
    content="Found 15 relevant papers on multi-agent reinforcement learning. Key themes: coordination, scalability, emergence."
)

# Planner receives the message
planner.perceive(research_results)

# Planner processes and responds
response = planner.act("Create action plan based on research findings")
print(f"\n{response.content}")

print(f"\nPlanner's message history: {len(planner.message_history)} messages")
```

## Key Design Principles

### 1. Single Responsibility
Each agent should have a clear, focused purpose.

### 2. Loose Coupling
Agents should be independent and communicate through well-defined interfaces.

### 3. Message Passing
Communication should be explicit via messages, not shared state.

### 4. Graceful Degradation
System should handle agent failures without complete collapse.

### 5. Observable Behavior
Agent actions and decisions should be trackable for debugging.
